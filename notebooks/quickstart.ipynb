### 🚀 Step 2: Run Live Cloud Extraction (`unitbytes/taobao-tmall-scraper`)
Grab your free Apify API token from **[Apify Console](https://apify.com/unitbytes/taobao-tmall-scraper?fpr=939u3w&fp_sid=gh_taobao_alibaba_colab)** (includes `$5.00` free monthly usage credit) and run the cell below. `getpass` prompts for your token securely so it is never stored in notebook outputs:


```python
!pip install -q apify-client pandas

from getpass import getpass
import pandas as pd
from apify_client import ApifyClient

APIFY_TOKEN = getpass("Enter your Apify API Token (from https://apify.com/unitbytes/taobao-tmall-scraper?fpr=939u3w&fp_sid=gh_taobao_alibaba_colab): ").strip()
if not APIFY_TOKEN:
    raise ValueError("Please enter a valid Apify API token to run live extraction.")

client = ApifyClient(APIFY_TOKEN)

run_input = {
    "queries": ["手机壳", "wireless earbuds"],
    "shopUrls": ["https://world.taobao.com/dianpu/100000001.htm"],
    "itemUrls": ["https://detail.tmall.com/item.htm?id=100000000001"],
    "scrapeDetails": True,
    "maxItems": 50,
    "sortBy": "sales"
}

print("Running unitbytes/taobao-tmall-scraper...")
run = client.actor("unitbytes/taobao-tmall-scraper").call(run_input=run_input)
items = list(client.dataset(run["defaultDatasetId"]).iterate_items())

df_live = pd.DataFrame(items)
print(f"Extracted {len(df_live)} live records!")
df_live.to_csv("taobao-tmall-scraper-results.csv", index=False)
df_live.head(10)
```

### ⚡ Step 3: Call the Synchronous `<500ms` Alibaba B2B REST API (RapidAPI Hub & Zyla API Hub)
Test the real-time synchronous `GET /v1/alibaba/search` endpoint on **[RapidAPI Hub](https://rapidapi.com/unitbytes/api/alibaba-b2b-wholesale-and-supplier-data)**:


```python
from getpass import getpass
import requests
import pandas as pd

RAPIDAPI_KEY = getpass("Enter your RapidAPI Key: ").strip()
resp = requests.get(
    "https://alibaba-b2b-wholesale-and-supplier-data.p.rapidapi.com/v1/alibaba/search",
    headers={"x-rapidapi-host": "alibaba-b2b-wholesale-and-supplier-data.p.rapidapi.com", "x-rapidapi-key": RAPIDAPI_KEY},
    params={"keyword": "wireless earbuds", "verified_only": "true", "trade_assurance_only": "true", "page": 1, "page_size": 5},
    timeout=30,
)
resp.raise_for_status()
pd.DataFrame(resp.json().get("data", {}).get("items", [])).head()
```
